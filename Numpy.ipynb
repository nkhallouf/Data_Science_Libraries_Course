## What is NumPy?

NumPy (or Numpy) is a Linear Algebra Library for Python, the reason it is so important for Data Science with Python is that almost all of the libraries in the PyData Ecosystem rely on NumPy as one of their main building blocks.
NumPy is a powerful Python library used for:
- Numerical computing
- Working with arrays
- Mathematical operations

It is the foundation for data science and machine learning.

## Using NumPy

Once you've installed NumPy you can import it as a library:

```python
import numpy as np
```

## Array Creation
NumPy provides multiple ways to create arrays depending on your needs.

### From a Python List

We can create an array by directly converting a list or list of lists:

```python
list = [1,2,3]
arr = np.array(list)
arr
```

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
np.array(matrix)
```

## Built-in Methods

There are lots of built-in ways to generate Arrays

### zeros and ones

Generate arrays of zeros or ones

```python
# Create an array filled with zeros (2 rows, 3 columns)
np.zeros((2, 3))
```

```python
# Create an array filled with ones
np.ones((2, 2))
```

### arange

Return evenly spaced values within a given interval.

```python
# Create values from 0 to 10 (exclusive) with step 2
np.arange(0, 10, 2)
```

### linspace
Return evenly spaced numbers over a specified interval.

```python
# Create 5 evenly spaced numbers between 0 and 1
np.linspace(0, 1, 5)
```

### Random 


```python
# Create a 2x2 array with random values between 0 and 1
np.random.rand(2, 2)
```

## Array Attributes

These properties help you understand the structure of your array.

```python
arr = np.array([[1, 2], [3, 4]])

# shape: dimensions of the array
# ndim: number of dimensions
# size: total number of elements
# dtype: data type of elements

arr.shape, arr.ndim, arr.size, arr.dtype
```

## Operations
NumPy allows element-wise operations and matrix operations.

```python
a = np.array([1, 2])
b = np.array([3, 4])

print(a + b)   # addition
print(a * b)    # multiplication
```

```python
# Matrix multiplication (dot product)
# Each row of the first matrix is multiplied by each column of the second matrix
np.dot(
    np.array([[1, 2], [3, 4]]),
    np.array([[5, 6], [7, 8]])
)
```

```python
# Basic statistics
arr = np.array([1, 2, 3])

total = np.sum(arr)  # total sum
print(total)
ave = np.mean(arr) # average
print(ave)
maximum = np.max(arr)  # maximum value
print(maximum)
minimum = np.min(arr)  # minimum value
print(minimum)
```

```python
# Standard deviation (spread of values)
# Standard deviation measures how spread out values are from the mean
# Low std → values are close to the mean
# High std → values are more spread out
np.std(np.array([1, 2, 3, 4, 5]))
```

## Indexing & Slicing

Access specific elements or parts of the array.

```python
arr = np.array([10, 20, 30])

arr[0]  # Access first element
```

```python
arr = np.array([10, 20, 30, 40])

arr[1:3]  # Slice elements from index 1 to 3 (exclusive)
```

```python
arr = np.array([[1, 2], [3, 4]])

arr[0, 1]  # Access element in 2D array (row 0, column 1)
```

```python
# Boolean indexing (filter values > 3)
arr = np.array([1, 5, 2, 8])

arr[arr > 3]
```

## Reshaping
Change the structure of arrays without changing the data.

```python
arr = np.arange(6)
print("arr = {}".format(arr))

reshape = arr.reshape(2, 3)  # Reshape array into 2 rows and 3 columns
print("reshape = {}".format(reshape))
```

```python
np.array([[1, 2], [3, 4]]).flatten()  # Flatten array into 1D
```

```python
np.array([[1, 2], [3, 4]]).T  # Transpose rows and columns
```
